```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("cleaned_marketing_campaign.xlsx")
```

Objective:
Analyze customer purchasing behavior and marketing campaign performance to identify which marketing channels and customer segments generate better ROI and campaign responses.

# **Visuals**

# Age Distribution

```python
sns.histplot(df["Age"], bins=20)
```

# Income Distribution

```python
sns.boxplot(y=df["Income"])
```

# Analyze Response

```python
sns.countplot(x="Response", data=df)
```

Most customers did not respond to the campaign, indicating opportunities for improving targeting strategies.

# **Channel Performance Analysis**

```python
channels = [
    "NumWebPurchases",
    "NumCatalogPurchases",
    "NumStorePurchases"
]

df[channels].sum().plot(kind="bar")
```

Web purchases contributed the highest transaction volume, suggesting digital channels are highly effective.

# **Product Spending Analysis**

```python
products = [
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds"
]

df[products].sum().sort_values().plot(kind="barh")
```

# **Income vs Spending**

```python
sns.scatterplot(
    x="Income",
    y="Total Spending",
    data=df
)
```

Higher-income customers tend to spend significantly more across product categories.

# **Correlation Heatmap**

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True), cmap="Blues")
```

# **Funnel Analysis**

```python
website_visits = df["NumWebVisitsMonth"].sum()

purchases = df["Total Purchases"].sum()

responses = df["Response"].sum()
```

```python
funnel_data = pd.DataFrame({
    "Stage": ["Website Visits", "Purchases", "Campaign Responses"],
    "Count": [website_visits, purchases, responses]
})

funnel_data
```

# Horizontal Bar Chart

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x="Count",
    y="Stage",
    data=funnel_data,
    palette="Blues_r"
)

plt.title("Marketing Funnel Analysis")
plt.xlabel("Customer Count")
plt.ylabel("Funnel Stage")

plt.show()
```

**Insight:**
The marketing funnel shows a significant drop from website visits to campaign responses. While customer website engagement is relatively high, only a smaller proportion converts into purchases and campaign acceptance. This suggests opportunities for improving targeting strategies and customer engagement campaigns.

# Conversion Rates

```python
purchase_conversion = (purchases / website_visits) * 100

response_conversion = (responses / purchases) * 100

print("Purchase Conversion Rate:", round(purchase_conversion, 2), "%")
print("Campaign Response Rate:", round(response_conversion, 2), "%")
```

# Recommendations

1. Increase investment in digital marketing channels since customers demonstrate strong purchasing behavior through online engagement.

2. Improve campaign personalization using customer demographics, purchasing behavior, and spending patterns to increase campaign response rates.

3. Focus promotional campaigns on high-spending customer segments to maximize return on marketing investment.
