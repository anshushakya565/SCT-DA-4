```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("marketing_campaign.xlsx")
```

```python
df.head()
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
income_mean = df['Income'].mean()
df['Income'] = df['Income'].fillna(income_mean)
```

```python
df.info()
```

```python
print(df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'])
```

```python
df['Age'] = 2025 - df['Year_Birth']
```

```python
df['Total Spending'] = df['MntFishProducts'] + df['MntFruits'] + df['MntGoldProds'] + df['MntMeatProducts'] + df['MntSweetProducts'] + df['MntWines']
```

```python
df['Total Purchases'] = df['NumCatalogPurchases'] + df['NumStorePurchases'] + df['NumWebPurchases']
```

```python
df.to_excel("cleaned_marketing_campaign.xlsx", index = False)
```
